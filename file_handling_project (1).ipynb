```python
import os
import csv
import json
```

```python
main_folder = "mainfolder"
```

```python
def load_data(main_folder):
  sales_data = {}
  product_details = {}
  product_descriptions = {}
  sales_file_path = os.path.join(main_folder, "sales_data.csv")
  with open(sales_file_path, mode='r') as file:
    reader = csv.reader(file)
    header = next(reader)
    for row in reader:
      sku = row[0]
      sales = list(map(int, row[1:]))
      sales_data[sku] = sales
      details_folder = os.path.join(main_folder, "product_details")
      for filename in os.listdir(details_folder):
        if filename.endswith(".json"):
          filepath = os.path.join(details_folder, filename)
          with open(filepath, 'r') as f:
            data = json.load(f)
            sku = filename.replace("details_", "").replace(".json", "")
            product_details[sku] = data
            descriptions_folder = os.path.join(main_folder, "product_descriptions")
            for filename in os.listdir(descriptions_folder):
              if filename.endswith(".txt"):
                filepath = os.path.join(descriptions_folder, filename)
                with open(filepath,"r") as f:
                  text = f.read().strip()
                  sku = filename.replace("description_", "").replace(".txt", "")
                  product_descriptions[sku] = text
  print ("Data loaded successfully!")
  return sales_data, product_details, product_descriptions




```

```python
sales_data, product_details, product_descriptions = load_data(main_folder)
print("Sales data items:", len(sales_data))
print("Product details items:", len(product_details))
print("Product descriptions items:", len(product_descriptions))
```

```python
def dump_data(mainfolder, sales_data, product_details, product_descriptions):
  sales_file_path = os.path.join(mainfolder, "sales_data.csv")
  with open(sales_file_path, mode="w", newline="")as file:
    writer = csv.writer(file)
    writer.writerow(["SKU", "Sale1", "Sale2", "Sale3", "Sale4", "Sale5"])
    for sku, sales in sales_data.items():
      writer.writerow([sku] + sales)
      details_folder = os.path.join(mainfolder, "product_details")
      for sku_id, details in product_details.items():
        filename = f"details_{sku_id}.json"
        filepath = os.path.join(details_folder, filename)
        with open(filepath, "w") as f:
          json.dump(details, f, indent=4)
          descriptions_folder = os.path.join(mainfolder, "product_descriptions")
          for sku_id, description in product_descriptions.items():
            filename = f"description_{sku_id}.txt"
            filepath = os.path.join(descriptions_folder, filename)
            with open (filepath, "w") as f:
              f.write(description)
              print("✔ All data saved successfully!")
```
